```python
import torch
import numpy as np
x = torch.tensor(6.7)
y= torch.tensor(1)
```

```python
rng = np.random.random()
```

```python
w = torch.tensor(rng, requires_grad=True)
b= torch.tensor(0.0, requires_grad= True)
```

```python
def binary_cross_entropy_loss(y_pred,y_true):
    epsilon = 1e-8
    predictions = torch.clamp(y_pred,epsilon,1-epsilon)
    loss = -(y_true * torch.log(y_pred) + (1-y_true) * torch.log(1-y_pred
    ))
    return loss
```

```python
z = w* x + b
y_pred = torch.sigmoid(z)
loss = binary_cross_entropy_loss(y_pred,y)
```

```python
print(loss)
```

```python
loss.backward()
```
